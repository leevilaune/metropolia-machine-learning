```python
hdf = pd.read_csv('health_v1.csv', sep=';', na_values=['',' '], decimal=',')
hdf
```

```python
hdf.dtypes
hdf['Sex'].unique()
```

```python
hdf['Exercise'] = hdf['Exercise'].astype('Int64')
hdf['Income'] = hdf['Income'].astype('Int64')
hdf['BirthYear'] = hdf['BirthYear'].astype('Int64')
hdf['Happiness'] = hdf['Happiness'].astype('Int64')
hdf['Sex'] = pd.Categorical(hdf['Sex'])
```

```python
hdf.dtypes
hdf['Sex']
```

```python
hdf['Sex'] = hdf['Sex'].astype(str)
hdf['Sex'] = hdf['Sex'].replace({'1': 'Male', '2':'Female'})
hdf['Sex'] = hdf['Sex'].astype('category')
```

```python
hdf['Sex'].unique()
```

```python
def plot_data(hdf):
    plt.figure(figsize=(20, 10))
    for i, col in enumerate(hdf.select_dtypes(include=['int', 'float'])):
        plt.subplot(3, 3, i + 1)
        hdf[col].plot(kind='hist', title=col)
        plt.xlabel(col)
```

```python
plot_data(hdf)
```

```python
def plot_data(hdf):
    plt.figure(figsize=(20,15))
    for i, col in enumerate(hdf.select_dtypes(include=['Int64', 'float64'])):
        plt.subplot(3,3,i+1)
        hdf[col].plot(kind='hist', title=col)
        plt.xlabel(col)
plot_data(hdf)        
```

```python
hdf.tail()
```

```python
fhdf = hdf
fhdf['Weight'] = fhdf['Weight'][(fhdf['Weight'] > 0) & (fhdf['Weight'] < 200)]
fhdf['BirthYear'] = fhdf['BirthYear'][(fhdf['BirthYear'] >= 1925) & (fhdf['BirthYear'] <= 2009)]
fhdf['Cholesterol'] = fhdf['Cholesterol'][fhdf['Cholesterol']>0]

fhdf.describe(include='all')
```

```python
plot_data(fhdf)
```

```python
lhdf = hdf
lhdf['Weight'] = lhdf['Weight'].apply(lambda x: x if 30 <= x <= 200 else None)
lhdf['Cholesterol'] = lhdf['Cholesterol'].apply(lambda x: x if x > 0 else None)

lhdf.describe()
```

```python
np.random.seed(42)
hdf['Group'] = np.random.choice(['A','B'], size=len(hdf))
hdf['Age'] = 2026-hdf['BirthYear']
hdf.rename(columns={'Smoking':'Smoker'}, inplace=True)
hdf
```

```python
corr = df.drop(columns=['ID','Sex']).corr()
plt.figure(figsize=(10,10))
plt.imshow(corr,cmap='coolwarm', interpolation='none')
plt.colorbar()
plt.xticks(range(len(corr)), corr.columns, rotation=90)
plt.yticks(range(len(corr)), corr.columns)
plt.show()
```

```python
x = np.linspace(0,2*np.pi,1000)
plt.figure(figsize=(8,8))
plt.subplot(3,2,1)
plt.plot(x,np.sin(x))
plt.title("Sin")

plt.subplot(3,2,2)
plt.plot(x,np.cos(x))
plt.title("Cos")

plt.subplot(3,2,3)
plt.plot(x,np.tan(x))
plt.title("Tan")

x = np.linspace(-1, 1, 1000)

plt.subplot(3,2,4)
plt.plot(x,np.arcsin(x))

plt.subplot(3,2,5)
plt.plot(x,np.arccos(x))

plt.subplot(3,2,6)
plt.plot(x,np.arctan(x))
plt.show()
```
