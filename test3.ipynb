```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree 

iris = pd.read_csv('datasets/iris.csv')
X = iris.drop(['species'], axis = 1)
y = iris['species'].astype('category')
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.20, random_state=69420)
print(X_train.shape, X_test.shape, y_train.shape, y_test.shape)
```

```python
model = DecisionTreeClassifier(max_depth=2, random_state=20)
model.fit(X_train, y_train)
```

```python
fig = plt.figure(figsize = (4, 4))
plot_tree(model, feature_names = X.keys(), class_names = ['setosa', 'versicolor', 'virginica'])
plt.show()
```

```python
from sklearn.metrics import confusion_matrix

preds = model.predict(X_test)
confusion_matrix(y_test, preds)
```
